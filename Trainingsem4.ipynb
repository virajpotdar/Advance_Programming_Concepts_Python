```python
# largest No
arr=[11,2,91,75,50]
largest=arr[0]
for i in arr:
  if i > largest:
    largest=i
print(largest)

# second largest
arr=[11,2,91,75,50]
largest=arr[0]
second_largest=arr[0]
for i in arr:
  if i > largest:
    second_largest=largest
    largest=i
  elif i > second_largest:
    second_largest=i
print(second_largest)
```

```python
# Move Zero to end
arr=[1,0,2,0,0,3]
insert_zeros=0
for i in range(len(arr)):
  if arr[i]!=0:
    arr[insert_zeros],arr[i]=arr[i],arr[insert_zeros]
    insert_zeros+=1
print(arr)
```

```python
# Rotate by one
arr=[1,2,3,4,5]
# output= [5,1,2,3,4]
last=arr[-1]
for i in range(len(arr)-1,0,-1):
  arr[i]=arr[i-1]
arr[0]=last
print(arr)
```

```python
# Calculate the Sum & Avg
# 10 20 30 40 50 60  then output- 150 30.0
arr=[10,20,30,40,50]

def calculate_sum_avg(arr):
  sum = 0
  for i in arr:
    sum+=i
  avg=sum/len(arr)
  return sum,avg

sum,avg=calculate_sum_avg(arr)
print(sum,avg)
```

```python
# Count even odd
arr=[1,2,3,4,5]
def count_even_odd(arr):
    even=0
    odd=0
    for i in arr:
        if i%2==0:
            even+=1
        else:
            odd+=1
    return even,odd
print(count_even_odd(arr))
```

```python
# Find Min and max in an given array
arr=[1,2,9,11,3]
def find_min_max(arr):
    min=arr[0]
    max=arr[0]
    for i in arr:
        if i < min:
            min=i
        if i > max:
            max=i
    return min,max
print(find_min_max(arr))
```

```python
# Remove duplicate characters
text = "Programming"
result = ""

for char in text:
    if char not in result:
        result += char

print(result)
```

```python
#character count
text = "banana"
counts = {}

for char in text:
    if char in counts:
        counts[char] += 1
    else:
        counts[char] = 1

print(counts)
```
